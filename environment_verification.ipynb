```python
import pandas as pd
import numpy as np
import matplotlib

print("pandas:", pd.__version__)
print("numpy:", np.__version__)
print("matplotlib:", matplotlib.__version__)
```

```python
import numpy as np

delays = np.array([5, 12, 3, 20, 7, 15, 2, 9])
print("Sample delay data (minutes):", delays)
print("Average delay:", np.mean(delays), "minutes")
print("Maximum delay:", np.max(delays), "minutes")
```

```python
import matplotlib.pyplot as plt

routes = ['Route A', 'Route B', 'Route C', 'Route D']
avg_delays = [12, 8, 20, 5]

plt.bar(routes, avg_delays, color='steelblue')
plt.title('Average Delay by Route (minutes)')
plt.ylabel('Delay (minutes)')
plt.show()
```
